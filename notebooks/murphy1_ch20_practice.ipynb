````python
# @title Solution (double-click to show the code)
import numpy as np
rng = np.random.default_rng(5)
X = rng.normal(size=(200, 5)) * np.array([4.0, 2.5, 0.8, 0.4, 0.2])
X -= X.mean(axis=0)
_, _, Vt = np.linalg.svd(X, full_matrices=False)
V2 = Vt[:2]
mse_pca = np.mean((X - (X @ V2.T) @ V2) ** 2)  # rank-2 PCA projection error
W1 = rng.normal(0, 0.2, (5, 2)); W2 = rng.normal(0, 0.2, (2, 5))
for _ in range(800):
    Z = X @ W1; Xh = Z @ W2; diff = (Xh - X) / len(X)
    W1, W2 = W1 - 0.08 * (X.T @ diff @ W2.T), W2 - 0.08 * (Z.T @ diff)
print(f"PCA rank-2 MSE:        {mse_pca:.5f}")
print(f"Linear AE rank-2 MSE:  {np.mean((X @ W1 @ W2 - X)**2):.5f}")
````

The trained linear autoencoder reaches the exact same minimum reconstruction MSE as rank-2 PCA to 5 decimal places: without non-linear activations, an L-unit bottleneck spans the exact same subspace as the top L principal eigenvectors (Murphy §20.3.1, Station 7).
