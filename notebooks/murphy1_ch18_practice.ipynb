````python
# @title Solution (double-click to show the code)
import numpy as np
x = np.linspace(-2, 2, 40); y = x ** 2
F = np.full_like(y, y.mean()); lr = 0.3
for m in range(1, 26):
    r = y - F  # negative gradient of ½(y - F)²
    best = min(x[:-1], key=lambda t: ((r[x <= t] - r[x <= t].mean())**2).sum() + ((r[x > t] - r[x > t].mean())**2).sum())
    stump = np.where(x <= best, r[x <= best].mean(), r[x > best].mean())
    F += lr * stump  # additive boosting update
    if m in (1, 5, 25): print(f"m = {m:2d} trees: MSE = {np.mean((y - F)**2):.4f}")
````

<details><summary><b>Answer and explanation</b></summary>

A single stump can only draw a 1-step staircase (MSE ≈ 0.97), but adding 25 boosted stumps fitted to successive residuals carves the parabola down to MSE ≈ 0.03—turning high-bias weak learners into an accurate predictor (Murphy §18.4).
