````python
# @title Solution (double-click to show the code)
import numpy as np
X = np.array([-2.2, -1.3, -0.4, 0.5, 1.4, 2.3])
y = np.sin(X)
ell, sig = 1.0, 0.1
rbf = lambda a, b: np.exp(-0.5 * (a[:, None] - b[None, :])**2 / ell**2)
K = rbf(X, X) + (sig**2) * np.eye(len(X))
X_test = np.array([0.0, 4.5])
K_s = rbf(X_test, X)
mu = K_s @ np.linalg.solve(K, y)  # posterior mean k*ᵀ (K + σ²I)⁻¹ y
var = 1.0 - np.sum(K_s * np.linalg.solve(K, K_s.T).T, axis=1)  # posterior variance
for x0, m, s in zip(X_test, mu, np.sqrt(var)):
    print(f"x* = {x0:4.1f}: posterior mean = {m:+.3f},  posterior std = {s:.3f}")
````

<details><summary><b>Answer and explanation</b></summary>

At x* = 0.0 (surrounded by training points) the posterior uncertainty is tiny (σ* ≈ 0.06), whereas at x* = 4.5 (far from data) the mean reverts toward 0 and the posterior std balloons back to nearly the full prior std 1.0 (σ* ≈ 0.996). That automatic 'I don't know far from data' is why GPs power Bayesian optimisation (Murphy §17.2, D2L ch. 18).
