````python
# @title Solution (double-click to show the code)
import numpy as np
rng = np.random.default_rng(12)
p_tilde = lambda z: 0.3 * np.exp(-(z + 2.0)**2) + 0.7 * np.exp(-(z - 2.0)**2)
z = 0.0; samples = []; acc = 0
for _ in range(20_000):
    zp = z + 1.8 * rng.normal()
    if rng.random() < p_tilde(zp) / p_tilde(z):  # Metropolis acceptance ratio (Z cancels!)
        z = zp; acc += 1
    samples.append(z)
post = np.array(samples[2000:])
print(f"accept rate: {acc/20000:.2f}   fraction in right mode (z>0): {np.mean(post > 0):.3f} (true 0.700)")
print(f"MCMC mean E[z]: {post.mean():.3f} (true +0.800)")
````

<details><summary><b>Answer and explanation</b></summary>

Even though we never computed the normalising constant Z, the chain spends 70% of its visits in the right mode (z > 0) and 30% in the left mode, matching the exact mixture weights and theoretical mean 0.3(−2) + 0.7(+2) = +0.80 (Murphy 2 §12.2).
