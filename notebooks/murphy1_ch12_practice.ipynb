````python
# @title Solution (double-click to show the code)
import numpy as np
rng = np.random.default_rng(5)
x = rng.normal(size=500); A = np.column_stack([np.ones(500), x])
y = rng.poisson(np.exp(0.5 + 0.8 * x))
w = np.zeros(2)
for _ in range(2000):
    w += 0.2 * A.T @ (y - np.exp(A @ w)) / len(y)  # observed minus expected, times the input (averaged)
print("estimated w:", w.round(3), "  true: [0.5, 0.8]")
````

<details><summary><b>Answer and explanation</b></summary>

The fit recovers weights close to the true [0.5, 0.8] (500 noisy counts can't pin them exactly). The gradient has the same 'observed − predicted' shape as linear and logistic regression. That's the exponential-family pattern from §3.4: every GLM's gradient is (y − E[y|x])·x, only the link (here exp) changes (Murphy §12.4).

Which output distribution and link would you use for: (a) house price, (b) whether a customer churns, (c) number of insurance claims per year, (d) which of 5 products a user clicks?

<details><summary><b>Hint</b></summary>

Match the distribution to what the target can be: any real number, yes/no, a count, one of K.

</details>

✍️ *Write your answer here (double-click to edit).*

<details><summary><b>Worked answer</b></summary>

(a) Gaussian with identity link: linear regression. (b) Bernoulli with logit link: logistic regression. (c) Poisson with log link, so the predicted rate is always positive. (d) Categorical with softmax: multinomial logistic regression. Same recipe each time: a linear score, then the squash that suits the output type (Murphy §12.2).

</details>
