## 2.1 · The COVID test by hand  
*Section §2.3 · derive*

A test has sensitivity P(+|sick) = 0.875 and specificity P(−|healthy) = 0.975. In a population where 10% are infected, what is P(sick | +)? Repeat for 1% infected.

<details><summary><b>Hint</b></summary>

P(+) = P(+|sick)P(sick) + P(+|healthy)P(healthy), and P(+|healthy) = 1 − specificity.

</details>

✍️ *Write your answer here (double-click to edit).*

<details><summary><b>Worked answer</b></summary>

With 10%: numerator = 0.875 × 0.1 = 0.0875. False positives = 0.025 × 0.9 = 0.0225. P(+) = 0.0875 + 0.0225 = 0.11. P(sick|+) = 0.0875 / 0.11 ≈ 0.795. With 1%: numerator = 0.00875, false positives = 0.025 × 0.99 = 0.02475, P(+) = 0.0335, P(sick|+) ≈ 0.261. Same test, very different meaning: when the disease is rare, most positives come from the large healthy group (Murphy §2.3.1).

</details>

````python
# @title Solution (double-click to show the code)
import numpy as np
z = np.array([1000.0, 1001.0, 1002.0])
with np.errstate(over="ignore", invalid="ignore"):
    naive = np.exp(z) / np.exp(z).sum()
print("naive :", naive)

def softmax(z):
    e = np.exp(z - z.max())  # subtract the largest score before exponentiating
    return e / e.sum()

def logsumexp(z):
    m = z.max()
    return m + np.log(np.exp(z - m).sum())  # pull the max out of the log
print("stable:", softmax(z).round(3))
print("logsumexp:", round(logsumexp(z), 4))
````

<details><summary><b>Answer and explanation</b></summary>

## 2.5 · Why σ′(z) = σ(z)(1 − σ(z))  
*Section §2.4 · derive*

Differentiate σ(z) = 1 / (1 + e^−z) and show the result can be written as σ(z)(1 − σ(z)).

<details><summary><b>Hint</b></summary>

Write σ = (1 + e^−z)^−1 and use the chain rule. Then notice 1 − σ = e^−z / (1 + e^−z).

</details>

✍️ *Write your answer here (double-click to edit).*

<details><summary><b>Worked answer</b></summary>

σ′(z) = −(1 + e^−z)^−2 · (−e^−z) = e^−z / (1 + e^−z)². Split it: [1 / (1 + e^−z)] · [e^−z / (1 + e^−z)] = σ(z) · (1 − σ(z)), because 1 − σ = (1 + e^−z − 1)/(1 + e^−z) = e^−z/(1 + e^−z). So the slope can be computed from the output alone, which is why backprop through a sigmoid is cheap (station 4).

</details>
