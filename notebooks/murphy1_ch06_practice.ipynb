````python
# @title Solution (double-click to show the code)
import numpy as np
def H(p):       return -np.sum(p * np.log2(p))  # average surprise
def CE(p, q):   return -np.sum(p * np.log2(q))  # surprise of p's outcomes under q
def KL(p, q):   return np.sum(p * np.log2(p / q))  # extra surprise
p = np.array([0.7, 0.1, 0.1, 0.1]); q = np.full(4, 0.25)
print(f"H(p) {H(p):.4f}  H(p,q) {CE(p,q):.4f}  H(p)+KL {H(p)+KL(p,q):.4f}")
print(f"KL(p||q) {KL(p,q):.4f}   KL(q||p) {KL(q,p):.4f}")
````

<details><summary><b>Answer and explanation</b></summary>

H(p) ≈ 1.357 bits and H(p, q) = 2 bits (a uniform code over 4 outcomes always costs 2 bits), so KL(p‖q) ≈ 0.643 = 2 − 1.357. Swapping the arguments gives KL(q‖p) ≈ 0.620: a different number. KL is not a distance, so which distribution is 'the truth' matters (Murphy §6.2.6).

<details><summary><b>Worked answer</b></summary>

0.9 × log₂ 1.8 + 0.1 × log₂ 0.2 = 0.9 × 0.848 + 0.1 × (−2.322) = 0.763 − 0.232 = 0.531 bits. Non-negativity: −KL = Σ p log(q/p) ≤ log Σ p (q/p) = log Σ q = log 1 = 0, by Jensen's inequality (log is concave, so the average of logs ≤ log of the average). Equality only when q = p (Murphy §6.2.4).

````python
# @title Solution (double-click to show the code)
import numpy as np
xs = np.array([-2, -1, 0, 1, 2]); ys = np.unique(xs ** 2)
J = np.zeros((5, len(ys)))
for i, x in enumerate(xs):
    J[i, list(ys).index(x ** 2)] = 1 / 5
px, py = J.sum(1), J.sum(0)
mi = sum(J[i, j] * np.log2(J[i, j] / (px[i] * py[j])) for i in range(5) for j in range(len(ys)) if J[i, j] > 0)  # sum over non-zero cells
corr = np.corrcoef(xs, xs ** 2)[0, 1]
print(f"correlation {corr:.4f}   mutual information {mi:.4f} bits   H(Y) {-(py*np.log2(py)).sum():.4f}")
````

## 6.4 · Why classifiers minimise cross-entropy  
*Section §6.2 · think*

Training a classifier minimises the average −log q(true label). Explain in two sentences why this is the same as making the model's distribution q match the data's distribution p.

<details><summary><b>Hint</b></summary>

Cross-entropy = entropy + KL. Which part can the model change?

</details>

✍️ *Write your answer here (double-click to edit).*

<details><summary><b>Worked answer</b></summary>

The average −log q over the data is the cross-entropy H(p, q) = H(p) + KL(p‖q). H(p) depends only on the data, so the only part training can lower is KL(p‖q), which is zero exactly when q = p. That's also maximum likelihood (Murphy §6.2.5).

</details>
