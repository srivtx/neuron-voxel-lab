## 1.1 · Zero training error  
*Section §1.2 · think*

You fit a degree-9 polynomial to 10 points and the training error is exactly 0. Is this a good model? What would you measure to find out?

<details><summary><b>Hint</b></summary>

How many knobs does a degree-9 polynomial have? How many points must it pass through?

</details>

✍️ *Write your answer here (double-click to edit).*

<details><summary><b>Worked answer</b></summary>

Not necessarily. A degree-9 polynomial has 10 coefficients, so it can pass through any 10 points exactly. Zero training error is guaranteed, which means it tells you nothing about the pattern. Measure the error on points the model never saw (a held-out test set, or cross-validation). If test error is much larger than training error, the model memorised the noise: that gap is overfitting (Murphy §1.2.3).

</details>

````python
# @title Solution (double-click to show the code)
import numpy as np
rng = np.random.default_rng(0)
f = lambda x: np.sin(2 * np.pi * x)
x_tr = rng.random(15); y_tr = f(x_tr) + rng.normal(0, 0.2, 15)
x_te = rng.random(200); y_te = f(x_te) + rng.normal(0, 0.2, 200)

best = None
for d in range(13):
    coef = np.polyfit(x_tr, y_tr, d)  # fit a degree-d polynomial to the training points
    tr = np.mean((np.polyval(coef, x_tr) - y_tr) ** 2)  # training mean squared error
    te = np.mean((np.polyval(coef, x_te) - y_te) ** 2)  # test mean squared error
    print(f"degree {d:2d}  train {tr:8.4f}  test {te:10.4f}")
    if best is None or te < best[1]:
        best = (d, te)
print("best degree on test data:", best[0])
````

<details><summary><b>Answer and explanation</b></summary>

Training error falls every time the degree goes up, because more knobs can always fit at least as well. Test error falls at first (the model can now bend like a sine), bottoms out around degree 3–6, then rises sharply as the curve starts chasing noise.

````python
# @title Solution (double-click to show the code)
import numpy as np
colours = ["red", "green", "blue", "green"]
cats = sorted(set(colours))                       # ['blue', 'green', 'red']
idx = [cats.index(c) for c in colours]
onehot = np.eye(len(cats))[idx]  # pick rows of the identity matrix
print(cats); print(onehot)

docs = ["the cat sat", "the dog sat", "the cat ate the fish"]
vocab = sorted({w for d in docs for w in d.split()})
bow = np.array([[d.split().count(w) for w in vocab] for d in docs])  # count each word in each sentence
print(vocab); print(bow)
````

<details><summary><b>Answer and explanation</b></summary>

One-hot gives each category its own column with a single 1, so a weighted sum can give each category its own weight without implying an order. Bag-of-words throws away word order and keeps counts: 'the cat ate the fish' has a 2 in the 'the' column.
